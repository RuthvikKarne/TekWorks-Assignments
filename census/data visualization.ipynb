```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
df=pd.read_csv("dataset.csv")
```

```python
df.head(10)
```

```python
plt.figure(figsize=(5,5))
sns.histplot(df['age'], bins=30, kde=True, color = 'green', edgecolor = 'black' ,alpha = 0.7)
plt.title('Distribution of Age')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.grid(True)
plt.xlim(0, 100)
plt.ylim(0, 50)
plt.show()
```

```python
plt.figure(figsize=(5,5))
sns.scatterplot(x=df['age'], y=df['chol'], data=df, color='blue', alpha=0.6)
plt.title('Age vs Cholesterol')
plt.xlabel('Age')
plt.ylabel('Cholesterol Level')
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(5,5))
sns.scatterplot(x=df['trestbps'], y=df['chol'], data=df, color='', alpha=0.6,marker='^')
plt.title('Age vs Cholesterol')
plt.xlabel('Age')
plt.ylabel('Cholesterol Level')
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(5,5))
sns.lineplot(x=df['age'], y=df['chol'], data=df, color='blue', marker='^')
plt.title('cp vs oldpeak')
plt.xlabel('cp')
plt.ylabel('oldpeak')
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(5,5))
sns.boxplot(y=df['thalach'] , color='green',showfliers=True , fill=True, linewidth=2.5)
plt.title('Box Plot of thalach')
plt.xlabel('thalach')
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(5,5))
plt.pie(df['target'].value_counts(),
labels=['no Disease','Disease'],
colors=['lightblue','lightcoral'],
autopct='%1.1f%%',
startangle=90,
shadow=True,
wedgeprops={'edgecolor':'Green','linewidth':1},
textprops={'fontsize':12,'color':'black'})
```

```python
from webbrowser import BackgroundBrowser


plt.figure(figsize=(5,5))
sns.barplot(x=df['sex'],y=df['age'],color="blue",alpha=0.7,width=0.5)
plt.show()
```

```python
plt.figure(figsize=(6,6))
cp_counts = df['cp'].value_counts()
cp_labels = ['Typical Angina (0)', 'Atypical Angina (1)', 'Non-anginal Pain (2)', 'Asymptomatic (3)']
plt.pie(cp_counts.values, 
        labels=[cp_labels[i] for i in cp_counts.index], 
        autopct='%1.1f%%',
        colors=['', 'lightcoral', 'lightgreen', 'lightskyblue'],
        startangle=90,
        shadow=True,
        wedgeprops={'edgecolor':'blue','linewidth':2},
        textprops={'fontsize':10})
plt.title('Chest Pain Type Distribution', fontsize=14, fontweight='bold')
plt.show()
```

```python
plt.figure(figsize=(6,4))
df.groupby("target")["age"].mean().plot(kind="bar")
plt.title("Average Age by Target")
plt.xlabel("Target")
plt.ylabel("Mean Age")
plt.show()
```

```python
data = df.groupby("sex")["target"].mean() * 100

labels = ["Female (0)", "Male (1)"]
values = data.values

plt.figure(figsize=(8,3))

colors = ["#ff6b6b", "#4dabf7"]   # different colors

plt.barh(labels, values, color=colors)

plt.title("Bullet Graph: Heart Disease % by Sex")
plt.xlabel("Heart Disease Percentage (%)")

# add value labels
for i, v in enumerate(values):
    plt.text(v + 1, i, f"{v:.1f}%", va="center")

plt.xlim(0, 100)
plt.show()
```
