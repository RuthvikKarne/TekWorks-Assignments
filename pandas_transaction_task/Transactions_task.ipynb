```python

import pandas as pd

# Load dataset
df = pd.read_csv('transactions.csv')

# Convert date column
df['t_date'] = pd.to_datetime(df['t_date'])

# Create Month and Quarter columns
df['Month'] = df['t_date'].dt.month
df['Month_Name'] = df['t_date'].dt.month_name()
df['Quarter'] = df['t_date'].dt.quarter

df.head()
```

```python
# 1.What is the total sales amount across all transactions?
total_sales= df['t_amt'].sum()
print(int(total_sales))
```

```python
#2.Which month had the highest total transaction amount? // trim the month from the date and then start the condition
month_highest=df.groupby('Month_Name')['t_amt'].sum().idxmax()
print(month_highest)
```

```python
# 3.What is the average transaction amount per customer?
avg_transaction=df.groupby('cust_id')['t_amt'].mean()
print(avg_transaction)
```

```python
# 4 What is the trend of total sales over months?
sales_over_months=df.groupby('Month')['t_amt'].sum().sort_index()
print(sales_over_months)
```

```python
#5.What is the highest single transaction amount recorded?

highest_single_transaction=df['t_amt'].max()
print(highest_single_transaction)
```

```python
# 6.Which service category contributes the most to total revenue?

service_category=df.groupby('services')['t_amt'].sum().idxmax()
print(service_category)
```

```python
# 7.Which product generated the highest revenue?
highest_revenue=df.groupby('products_used')['t_amt'].sum().idxmax()
print(highest_revenue)
```

```python
# 8.How does the average transaction amount vary between services?
avg_transaction_service=df.groupby('services')['t_amt'].mean()
print(avg_transaction_service)
```

```python
#9.How many unique customers made purchases?
cust_unique=df['cust_id'].nunique()
print(cust_unique)
```

```python
# 10.Which customers have spent the most overall?
customer_spend=df.groupby('cust_id')['t_amt'].sum().sort_values(ascending=False).head()
customer_spend
```

```python
# 11.What is the average number of transactions per customer?
avg_transaction_customer=df.groupby('cust_id').size().mean()
int(avg_transaction_customer)
```

```python
# 12.Are there customers who made multiple purchases in different categories?
multiple_categories=df.groupby('cust_id')['services'].nunique().loc[lambda x: x > 1]
multiple_categories
```

```python
# 13.What percentage of customers are repeat buyers?
repeated_buyers=(df.groupby('cust_id').size().gt(1).sum() / df['cust_id'].nunique()) * 100
print(repeated_buyers)
```

```python
# 14.Which product category has the highest total sales?
product_with_highest_sales=df.groupby('products_used')['t_amt'].sum().idxmax()
print(product_with_highest_sales)
```

```python
# 15.Which services are most popular by transaction count?
popular_service=df['services'].value_counts().idxmax()
print(popular_service)
```

```python
# 16.For each service, which product type is purchased most frequently?
frequest_profuct=df.groupby(['services','products_used']).size().reset_index(name='Count').sort_values(['services','Count'], ascending=[True, False]).groupby('services').first()
print(frequest_profuct)
```

```python
# 17.What is the average transaction amount per product type?
avg_transaction=df.groupby('products_used')['t_amt'].mean()
print(avg_transaction)
```

```python
# 18.Are there specific services where customers spend significantly more?
services_more_spend=df.groupby('services')['t_amt'].mean().sort_values(ascending=False)
print(services_more_spend)
```

```python
# 19.Which state has the highest total sales?
highest_sales_state=df.groupby('state')['t_amt'].sum().idxmax()
print(highest_sales_state)
```

```python
# 20.Which city records the highest number of transactions?
city_highesttransaction=df['city'].value_counts().idxmax()
print(city_highesttransaction)
```

```python
# 21.What is the average spending per transaction in each state?
avg_spending_state=df.groupby('state')['t_amt'].mean()
print(avg_spending_state)
```

```python
# 22.Are certain services more popular in specific states?
services_per_states=df.groupby(['state','services']).size().reset_index(name='Count').sort_values('Count', ascending=False)
print(services_per_states)
```

```python
# 23.Which states buy the most Outdoor Recreation products?
outdoor_recreation=df[df['products_used']=='Outdoor Recreation'].groupby('state')['t_amt'].sum().sort_values(ascending=False)
print(outdoor_recreation)
```

```python
# 24.Compare average spending between California and Texas customers.
compare=df[df['state'].isin(['California','Texas'])].groupby('state')['t_amt'].mean()
print(compare)
```

```python
# 25.Which quarter of the year saw the highest sales?
quarter_highestsale=df.groupby('Quarter')['t_amt'].sum().idxmax()
print(quarter_highestsale)
```

```python
# 26.How does total sales vary month by month?
sales_monthh=df.groupby('Month')['t_amt'].sum().sort_index()
print(sales_monthh)
```

```python
# 27.What is the total number of transactions per month?
transaction_month=df.groupby('Month').size()
print(transaction_month)
```

```python
# 28.Is there a particular season when sports equipment sales spike?
sport_month=df[df['products_used']=='Sports Equipment'].groupby('Month')['t_amt'].sum()
sport_month
```

```python
# 29.How many transactions were done using credit?
credit_transaction=df[df['t_details']=='credit'].shape[0]
print(credit_transaction)
```

```python
# 30.What is the total revenue generated from credit transactions?
revenue_credit_transaction=df[df['t_details']=='Credit']['t_amt'].sum()
print(revenue_credit_transaction)
```

```python
# 31.Is there a difference in average spending between credit and debit customers?
avg_spending=df.groupby('t_details')['t_amt'].mean()
print(avg_spending)
```

```python
# 32.Which states or cities should the company focus on for marketing high-value products?
highvalueMarketing=df.groupby('state')['t_amt'].sum().sort_values(ascending=False).head()
print(highvalueMarketing)
```

```python
# 33.Should more inventory be stocked for Exercise & Fitness products given their popularity?
exercise_fitness=df[df['products_used']=='Exercise & Fitness'].shape[0]
print(exercise_fitness)
```

```python
# 34.Are there product categories with high sales but low average transaction amounts?
category_sales=df.groupby('products_used')['t_amt'].agg(['sum','mean'])
print(category_sales)
```

```python
# 35.Are there underperforming service categories that need promotional offers?
underperform=df.groupby('services')['t_amt'].sum().sort_values()
underperform
```
